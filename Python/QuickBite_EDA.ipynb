```python
#Gross Revenue
gross_rev = df_orders[df_orders['is_cancelled']=='N']
gross_rev['subtotal_amount'].sum().round(2)
```

Gross revenue was Rs. 4,68,93,793.

```python
#Net Revenue
net_rev = gross_rev['total_amount'].sum()
net_rev
```

Net revenue was Rs. 4,85,61,115.

```python
#Total Discounts
total_discounts = gross_rev['discount_amount'].sum()
total_discounts.round(2)
```

Total Discounts offered was Rs. 28,17,982.

```python
#Delivery Revenue
delivery_rev = gross_rev['delivery_fee'].sum()
delivery_rev
```

Delivery revenue was Rs. 44,85,305

```python
#AOV (Average Order Value)
aov = gross_rev['subtotal_amount'].sum()/gross_rev['order_id'].count()
aov.round(2)
```

Average Order value was Rs. 339.68.

```python
#Revenue by Month
gross_rev.groupby(gross_rev['order_timestamp'].dt.month_name())['subtotal_amount'].sum().sort_values(ascending = False)
```

The highest revenue was generated in March while lowest was generated in August.

```python
#Revenue by city
orders_restaurant = pd.merge(left=df_restaurant, right=gross_rev, how= 'inner', on='restaurant_id')
orders_restaurant.groupby(orders_restaurant['city'])['subtotal_amount'].sum().sort_values(ascending = False)
```

The highest revenue was generated in Bengaluru while lowest was generated in Pune city.

```python
#Revenue by Restaurant
orders_restaurant.groupby(['restaurant_id', 'restaurant_name'])['subtotal_amount'].sum().sort_values(ascending = False)
```

The Hot & Crispy Pizza Bhojanalay restaurant by restaurant id REST16780 earned the most Rs. 8170.11 while Annapurna Darshini Darbar by restaurant id earned lowest Rs. 121.06 amont 19971 restaurants.


```python
#Revenue by Cuisine
orders_restaurant.groupby(['cuisine_type'])['subtotal_amount'].sum().sort_values(ascending=False)
```

Among 8 cuisine types, North Indian cuisine generated most revenue Rs. 92,70,090.55 while Healthy food generated the least Rs. 23,75,681.92.

```python
#Top 10 restaurants by revenue
#orders_restaurant.groupby(['restaurant_id','restaurant_name']).agg({'subtotal_amount': 'sum'}).sort_values(by = 'subtotal_amount',ascending = False).head(5)
orders_restaurant.groupby(['restaurant_id','restaurant_name'])['subtotal_amount'].sum().sort_values(ascending = False).head(10)
```

Top 10 restaurants by revenue were namely Hot & Crispy Pizza Bhojanalay, Punjabi Darshini Palace, Punjabi Wraps Bhojanalay, Hot & Crispy Paratha Zone, Thindi Mane Grill Palace, Spicy Delights Bhojanalay, Thindi Mane House Hub, Fresh Thali Adda, Bombay Kitchen Heaven and Punjabi Curry Delight.

```python
#Top 10 restaurants by orders
orders_restaurant.groupby(['restaurant_id','restaurant_name'])['order_id'].count().sort_values(ascending = False).head(10)
```

The top 10 restaurants by orders were Hot & Crispy Pizza Bhojanalay, Thindi Mane Grill Palace, Punjabi Darshini Palace, Thindi Mane House Hub, Punjabi Wraps Bhojanalay, Fresh Thali Adda, Hot & Crispy Paratha Zone, Hot & Crispy Express Darbar, Grand Express Bhojanalay and Flavours of Delights Darbar.

```python
#Top 10 restaurants by AOV
aov_df = orders_restaurant.groupby(['restaurant_id','restaurant_name']).agg(total_rev = ('subtotal_amount', 'sum'), total_orders = ('order_id', 'count')).reset_index()
aov_df['aov'] = aov_df['total_rev']/aov_df['total_orders']
aov_df.sort_values(by='aov',ascending=False).head(10)
```

```python
#Revenue contribution %
rev_rest = orders_restaurant.groupby(['restaurant_id','restaurant_name'])['subtotal_amount'].sum()
total_rev_rest = rev_rest.sum()
rev_df = (rev_rest*100)/total_rev_rest
rev_df
```

```python
#cancellation Rate
order_rest_all = pd.merge(left=df_orders, right=df_restaurant, on='restaurant_id', how='inner')
order_rest_all['cancel_flag'] = (order_rest_all['is_cancelled']=='Y')
order_rest_all
cancel_rest = order_rest_all.groupby(['restaurant_id','restaurant_name']).agg(total_cancel = ('cancel_flag','sum'), total_orders = ('order_id','count')).reset_index()
cancel_rest['cancellation_rate'] = (cancel_rest['total_cancel']*100)/cancel_rest['total_orders']
cancel_rest
```

<b> LEVEL 3: Advanced Pandas <b>

Assignment 7: Customer Retention

```python
#Customer Lifespan

cust_orders = pd.merge(left=df_customer, right=df_orders, on='customer_id', how='inner')
#cust_orders.head()
cust_lifespan = cust_orders.groupby(['customer_id']).agg(first_order = ('order_timestamp','min'), last_order = ('order_timestamp','max'))
cust_lifespan['lifespan'] = cust_lifespan['last_order']-cust_lifespan['first_order']
cust_lifespan.sort_values(by='lifespan', ascending=False)
```

Customer with most lifespan was 271 days while lowest was 0 days.
